```python
###creating a txt file

import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
#Here we are making a txt files using python

sample_texts = {
    "../data/text_files/python.txt": "This is the content of python.txt.",

    "../data/text_files/machine_learning.txt": "This is the content of machine_learning.txt.",
}


for file_path, content in sample_texts.items():
    with open(file_path, "w", encoding = "utf-8") as f:
        f.write(content)

```

```python
##textloader = here we are converting the txt files into the langchain documents


from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python.txt", encoding = "utf-8")
document = loader.load()
print(document)
```

```python
##Directory loader: Here we will be converting a specific kind of files present in a specific directory

from langchain_community.document_loaders import DirectoryLoader

##Load all the txt files in the directory

dir_loader = DirectoryLoader("../data/text_files", ## specifying the address of the directory
glob = "**/*.txt", ## specifying the type of files to choose from inside the directory
loader_cls = TextLoader, ## depending upon the type of files we're converting, we will put that loader here
loader_kwargs = {"encoding":"utf-8"}, 
show_progress = False
)

document = dir_loader.load()
print(document)
```

```python
## Now we will learn to load a pdf file as well
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

pdf_loader = DirectoryLoader("../data/pdf", ## specifying the address of the directory
glob = "**/*.pdf", ## specifying the type of files to choose from inside the directory
loader_cls = PyMuPDFLoader, ## depending upon the type of files we're converting, we will put that loader here 
show_progress = False
)

doc = pdf_loader.load()
doc
```
